### 3. Statistical Analysis – Pearson Correlation

This section investigates the statistical relationships between life evaluation and six explanatory factors.

Pearson's correlation coefficient is used to measure the strength and direction of linear relationships. P-values are also calculated to assess statistical significance.

A significance level of 0.05 is used. A p-value below 0.05 indicates statistical evidence of a linear association, but it does not imply causation.

```python
from scipy.stats import pearsonr

happiness_column = "Life evaluation (3-year average)"

results = []

for column, label in factor_names.items():

    valid_data = df_factors_2025[
        [happiness_column, column]
    ].dropna()

    r_value, p_value = pearsonr(
        valid_data[happiness_column],
        valid_data[column]
    )

    results.append({
        "Factor": label,
        "Correlation": r_value,
        "P-value": p_value,
        "Number of countries": len(valid_data)
    })

correlation_results = pd.DataFrame(results)

correlation_results = correlation_results.sort_values(
    by="Correlation",
    ascending=False
)

display(correlation_results.round(4))
```

```python
correlation_results["Statistically significant"] = (
    correlation_results["P-value"] < 0.05
)

display(correlation_results)
```

```python
number_of_tests = len(correlation_results)

correlation_results["Bonferroni adjusted p-value"] = (
    correlation_results["P-value"] * number_of_tests
).clip(upper=1)

correlation_results["Significant after correction"] = (
    correlation_results["Bonferroni adjusted p-value"] < 0.05
)

display(
    correlation_results[
        [
            "Factor",
            "Correlation",
            "P-value",
            "Bonferroni adjusted p-value",
            "Significant after correction"
        ]
    ].round(4)
)
```

### 4. Interpretation of Statistical Results

The Pearson correlation analysis reveals positive associations between life evaluation and five of the six explanatory factors.

Social Support shows the strongest correlation with life evaluation (r = 0.8054), followed by GDP per Capita (r = 0.7446), Healthy Life Expectancy (r = 0.6775), and Freedom to Make Life Choices (r = 0.6398).

Perceptions of Corruption presents a moderate positive correlation (r = 0.3925). Generosity shows a very weak correlation (r = 0.0419) and is not statistically significant (p = 0.6147).

After applying the Bonferroni correction for six statistical tests, five factors remain statistically significant, while Generosity does not.

These results indicate statistical associations, not causal relationships. They should also be interpreted in light of the dataset's methodology: the explanatory variables are reported as contributions to the happiness score, rather than necessarily being raw measurements of the underlying social and economic conditions.

```python
yearly_correlations = []

for year in sorted(df_factors["Year"].unique()):

    yearly_data = df_factors[df_factors["Year"] == year]

    for column, label in factor_names.items():

        valid_data = yearly_data[
            [happiness_column, column]
        ].dropna()

        if len(valid_data) >= 3:
            r_value, p_value = pearsonr(
                valid_data[happiness_column],
                valid_data[column]
            )

            yearly_correlations.append({
                "Year": year,
                "Factor": label,
                "Correlation": r_value,
                "P-value": p_value,
                "Number of countries": len(valid_data)
            })

yearly_correlations_df = pd.DataFrame(yearly_correlations)

display(
    yearly_correlations_df
    .sort_values(["Factor", "Year"])
    .round(3)
)
```
