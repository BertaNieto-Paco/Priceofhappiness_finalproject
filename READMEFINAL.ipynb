# The Price of Happiness 🌍

## Exploring the Factors Associated with Happiness Across Countries

### 📌 Project Overview

Why are some countries happier than others? Is happiness mainly related to economic prosperity, or do social factors play an equally important role?

**The Price of Happiness** is an end-to-end data analytics project that explores happiness levels across countries and their relationship with different social, economic and geographical factors.

Using data from the World Happiness Report, this project analyses happiness trends over time, compares countries and regions, and investigates the statistical relationships between happiness scores and six explanatory components.

The main objective is to identify patterns and understand which factors show the strongest association with reported happiness.

### 🎯 Research Questions

This project aims to answer the following questions:

- How have happiness scores changed over time?
- Which countries have the highest and lowest happiness scores?
- Are higher GDP per capita levels associated with higher happiness scores?
- How is social support related to happiness?
- Is healthy life expectancy associated with happiness?
- Is freedom to make life choices related to happiness?
- Are there significant differences between geographical regions?
- Which explanatory components show the strongest statistical association with happiness?

### 📊 Datasets

This project uses two datasets:

**1. World Happiness Dataset**

The main dataset contains 2,116 observations and 28 original columns, covering the years from 2011 to 2025.

It includes:
- Country and year
- Happiness ranking
- Life evaluation scores
- GDP per capita component
- Social support component
- Healthy life expectancy component
- Freedom to make life choices component
- Generosity component
- Perceptions of corruption component

**2. Countries and Geographic Regions Dataset**

A second dataset was used to match countries with their geographical information.

It contains:
- Country names
- ISO country codes
- Geographic regions
- Geographic sub-regions

This dataset allowed the project to include a regional and sub-regional comparison.

### 🔍 Project Workflow

The project follows an end-to-end data analytics workflow:

**1. Data Collection**

Importing the World Happiness dataset and the geographical information dataset.

**2. Data Cleaning and Preparation**

- Removing empty columns.
- Converting numerical values into the correct data types.
- Checking missing values and duplicated records.
- Preparing separate datasets for different analyses.
- Matching country names across datasets.

**3. SQL Database**

- Creating a SQLite database.
- Storing the cleaned datasets in SQL tables.
- Writing queries to calculate averages and rankings.
- Comparing SQL results with Python calculations.

**4. Exploratory Data Analysis (EDA)**

- Analysing happiness trends over time.
- Comparing the highest and lowest-ranking countries.

- Examining the distribution of happiness scores.

**5. Statistical Analysis**

- Calculating Pearson correlation coefficients.
- Analysing p-values and statistical significance.

**6. Data Visualisation**

Creating charts and heatmaps to communicate the main findings.

**7. Conclusions**

Interpreting the results, identifying limitations and suggesting possible future improvements.

### 📈 Key Findings

The main findings of the analysis are:

**Happiness Trends**

- Average happiness increased from 5.392 in 2011 to 5.650 in 2025.
- Median happiness increased from 5.230 to 5.930.

**Country Comparison (2025)**

- Finland ranked first, with a happiness score of 7.764.
- Afghanistan ranked last, with a score of 1.446.
- The difference between the two countries was 6.318 points.

**Correlation Analysis (2025)**

The strongest positive correlation was found for social support, followed by GDP per capita.

| Factor | Pearson Correlation |
|---|---:|
| Social Support | 0.805 |
| GDP per Capita | 0.745 |
| Healthy Life Expectancy | 0.678 |
| Freedom | 0.640 |
| Perceptions of Corruption | 0.392 |
| Generosity | 0.042 |

Generosity showed the weakest association and was not statistically significant.

**Regional Analysis (2025)**

- Oceania had the highest average happiness score (6.956).
- Europe followed with an average of 6.531.
- Africa had the lowest average score (4.480).

### ⚠️ Limitations

Some limitations should be considered when interpreting the results:

- The number of countries included varies between years.
- The dataset does not contain observations for 2013.
- Some explanatory variables are only available from 2019 onwards.
- Oceania includes only two countries in the 2025 regional analysis.
- The explanatory factors are components used in the World Happiness Report's score decomposition, rather than independent raw social and economic indicators.
- Correlation does not imply causation.

Therefore, the results identify statistical associations and patterns, but do not establish cause-and-effect relationships.

### 💡 Conclusion

The analysis suggests that happiness is associated with several economic and social factors.

Social support showed the strongest positive correlation with happiness in 2025, followed by GDP per capita, healthy life expectancy and freedom.

The project also revealed considerable differences between countries and geographical regions.

Overall, this project demonstrates how Python, SQL, statistical analysis and data visualisation can be combined to explore a real-world question through an end-to-end data analytics process.
